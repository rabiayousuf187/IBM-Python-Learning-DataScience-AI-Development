```python
import os
import urllib.request
import pandas as pd

# 1. Exact IBM Cloud storage link to your product sales data
csv_url = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/LXjSAttmoxJfEG6il1Bqfw/Product-sales.csv"
local_filename = "Product-sales.csv"

print(f"📡 Requesting download from: {csv_url}")

async def download(url, filename):
    try:
        # 2. Safely download and write the file locally
        urllib.request.urlretrieve(url, local_filename)
        print("✅ File successfully downloaded!")

        # 3. Read and display the CSV dataset
        # df = pd.read_csv(local_filename)
        # display(df.head())

    except Exception as e:
        print(f"❌ Download failed: {e}")

await download(csv_url, local_filename)

# 3. Read the downloaded target file
if os.path.exists(local_filename):
    df = pd.read_csv(local_filename)
    display(df.head())
```
